Maintained command wrapper, added on 2026-10-02; not the original successful-run notebook. Run from the repository root. Install requirements-mt5.txt and a suitable CUDA PyTorch build. It loads the frozen exports directly and verifies their hashes. Training and generation require a GPU and can be expensive. The maintained runner differs from the preserved successful run in evaluation/checkpoint frequency; see pretrained_baseline/README.md.

```python
from pathlib import Path
import subprocess, sys
assert Path("pretrained_baseline/run_baseline.py").exists(), "Set the working directory to the repository root."
def run_stage(*args):
    subprocess.run([sys.executable, "pretrained_baseline/run_baseline.py", *args], check=True)
```

## Training

Uses all 6,436 training passages and the 804 validation passages for evaluation/model selection. Selectively execute this stage; no automatic retraining is required to inspect saved evidence.

```python
run_stage("train", "--output", "runs/mt5-full")
```

Generate predictions for the 805 test passages from the selected checkpoint. Test targets do not enter model fitting.
